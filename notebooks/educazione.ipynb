````python
snapshot = youth[(youth["territorio_nome"].eq("Bagheria")) & (youth["anno"].eq(2024))].iloc[0]
components = [
    ("Studenti", snapshot["quota_studenti"], "#2F6BFF"),
    ("Occupati", snapshot["quota_occupati"], "#2A9D8F"),
    ("In cerca", snapshot["quota_in_cerca"], "#725AC1"),
    ("Inattivi non studenti", snapshot["quota_inattivi_non_studenti"], "#F28E2B"),
]
composition_total = sum(value for _, value, _ in components)
assert np.isclose(composition_total, 100.0, atol=1e-8)

fig, ax = plt.subplots(figsize=(14, 4.8))
left = 0.0
for label, value, color in components:
    ax.barh(0, value, left=left, height=0.46, color=color)
    ax.text(
        left + value / 2, 0, f"{label}\n{fmt(value)}%",
        ha="center", va="center", color="white", fontweight="bold",
        fontsize=9 if value < 10 else 10.5,
    )
    left += value

outside_start = 100 - snapshot["quota_fuori_lavoro_studio"]
ax.annotate(
    "", xy=(100, -0.40), xytext=(outside_start, -0.40),
    arrowprops={"arrowstyle":"<->", "color":"#D1495B", "linewidth":1.8},
)
ax.text(
    (outside_start + 100) / 2, -0.55,
    f"Fuori da lavoro e studio: {fmt(snapshot['quota_fuori_lavoro_studio'])}%\n"
    f"= {fmt(snapshot['quota_in_cerca'])}% in cerca + {fmt(snapshot['quota_inattivi_non_studenti'])}% inattivi",
    ha="center", va="top", color="#D1495B", fontweight="bold", fontsize=10,
)
ax.set_xlim(0, 100)
ax.set_ylim(-0.85, 0.55)
ax.set_yticks([])
ax.set_xlabel("composizione dei residenti 15-24 (%)")
ax.set_xticks([0, 20, 40, 60, 80, 100])
ax.grid(axis="x", color="#E4EAF1", linewidth=0.8)
for spine in ["top", "right", "left"]:
    ax.spines[spine].set_visible(False)
ax.set_title(
    "Bagheria 2024: quattro condizioni che sommano esattamente al 100%",
    loc="left", fontsize=18, fontweight="bold", color="#102A43", pad=28,
)
ax.text(
    0, 1.02,
    f"{count(snapshot['popolazione'])} residenti 15-24 · il 26,9% è un aggregato delle ultime due categorie, non una quinta categoria",
    transform=ax.transAxes, fontsize=10, color="#52606D",
)
fig.subplots_adjust(left=0.03, right=0.98, top=0.76, bottom=0.27)
fig.text(0.125, 0.02, "Fonte: elaborazione su IstatData. Eventuali scarti visivi dipendono solo dall'arrotondamento a una cifra decimale.", fontsize=8.5, color="#52606D")
show_figure(fig, "Composizione completa e mutuamente esclusiva dei giovani 15-24 di Bagheria nel 2024")
````

### Grafico 1 — Che cosa è cambiato davvero tra 1991 e 2011

<div class="read-guide">
  <div class="label">Domanda</div><div class="text">Quali indicatori migliorano e con quale intensità?</div>
  <div class="label">Lettura</div><div class="text">Il punto grigio è il 1991, quello blu il 2011. Accanto a ogni indicatore è specificato se è favorevole salire o scendere.</div>

````python
h = history.set_index("indicatore")
order = ["I6", "I7", "I5", "L4", "L14"]
labels = {
    "I6":"Diploma/laurea 25-64  ↑ meglio",
    "I7":"Titolo universitario 30-34  ↑ meglio",
    "I5":"Uscita precoce 15-24  ↓ meglio",
    "L4":"NEET 15-29  ↓ meglio",
    "L14":"Occupazione 15-29  ↑ meglio",
}

fig, ax = plt.subplots(figsize=(14, 6.4))
y_positions = np.arange(len(order))[::-1]
for y, indicator in zip(y_positions, order):
    start = h.loc[indicator, "valore_1991"]
    end = h.loc[indicator, "valore_2011"]
    favorable_change = (end - start) if indicator in {"I6", "I7", "L14"} else (start - end)
    ax.plot([start, end], [y, y], color="#CBD5E1", linewidth=4, solid_capstyle="round", zorder=1)
    ax.scatter(start, y, s=115, color="#AAB2BD", edgecolor="white", linewidth=1.5, zorder=2)
    ax.scatter(end, y, s=130, color="#2F6BFF", edgecolor="white", linewidth=1.5, zorder=3)
    ax.annotate(f"1991  {fmt(start)}%", (start, y), xytext=(0, 12), textcoords="offset points", ha="center", color="#627D98", fontsize=9)
    ax.annotate(f"2011  {fmt(end)}%", (end, y), xytext=(0, -17), textcoords="offset points", ha="center", color="#102A43", fontsize=9, fontweight="bold")
    ax.text(53, y, f"+{fmt(favorable_change)} p.p. favorevoli", va="center", color="#2A9D8F", fontweight="bold", fontsize=10)

ax.set_yticks(y_positions, [labels[indicator] for indicator in order])
ax.set_xlim(0, 67)
ax.set_ylim(-0.55, 4.55)
ax.set_xlabel("quota osservata (%)")
ax.grid(axis="x", color="#E4EAF1", linewidth=0.8)
for spine in ["top", "right", "left"]:
    ax.spines[spine].set_visible(False)
ax.set_title(
    "1991-2011: tutti gli indicatori migliorano, ma con intensità molto diverse",
    loc="left", fontsize=18, fontweight="bold", color="#102A43", pad=30,
)
ax.text(
    0, 1.01,
    "Confronto diretto delle quote: il cambiamento più forte è educativo; occupazione e NEET si muovono poco.",
    transform=ax.transAxes, fontsize=10, color="#52606D",
)
fig.subplots_adjust(left=0.25, right=0.98, top=0.79, bottom=0.17)
fig.text(0.125, 0.02, "Fonte: Istat 8milaCensus. Le fasce d'età sono indicate nei nomi; gli indicatori non descrivono una singola coorte.", fontsize=8.5, color="#52606D")
show_figure(fig, "Confronto diretto tra i valori 1991 e 2011 degli indicatori storici di Bagheria")

display(Markdown(f'''
**Cosa vediamo.** L'uscita precoce 15-24 scende dal **{fmt(h.loc['I5','valore_1991'])}%** al
**{fmt(h.loc['I5','valore_2011'])}%**; diploma/laurea 25-64 sale dal
**{fmt(h.loc['I6','valore_1991'])}%** al **{fmt(h.loc['I6','valore_2011'])}%** e il titolo
universitario 30-34 dal **{fmt(h.loc['I7','valore_1991'])}%** al
**{fmt(h.loc['I7','valore_2011'])}%**. Il miglioramento educativo è consistente. NEET e
occupazione giovanile cambiano invece poco: NEET dal **{fmt(h.loc['L4','valore_1991'])}%** al
**{fmt(h.loc['L4','valore_2011'])}%**, occupazione dal **{fmt(h.loc['L14','valore_1991'])}%**
al **{fmt(h.loc['L14','valore_2011'])}%**.
'''))
````

## Ponte 19 Bagheria

**Un servizio comunale di transizione e riattivazione 18-24.** Il target operativo 18-24 è
più stretto del perimetro statistico 15-24: è una scelta di implementazione, non un numero

| Evidenza osservata | Scelta di progetto | Perché |
|---|---|---|
| 19,0% inattivi non studenti; 70,6% degli esclusi | Outreach e case management | chi non cerca non arriva spontaneamente allo sportello |
| Gap occupazionale ancora −3,1 p.p. | KPI di convergenza e tenuta | contare gli utenti non dimostra il recupero territoriale |

| Nessun microdato pubblico titolo × condizione | Dataset longitudinale del servizio | misurare finalmente titolo, azione ed esito individuale |

### Come funziona il servizio

<div class="pipeline">
  <div class="pipeline-step"><div class="step">01</div><h4>Intercettare</h4><p>Ultimo anno scolastico, interruzioni e segnalazioni volontarie.</p></div>
  <div class="pipeline-step"><div class="step">02</div><h4>Raggiungere</h4><p>Outreach verso chi è fuori e non cerca attivamente.</p></div>
  <div class="pipeline-step"><div class="step">03</div><h4>Diagnosticare</h4><p>Titolo, indirizzo, durata dell'inattività, barriera e obiettivo.</p></div>
  <div class="pipeline-step"><div class="step">04</div><h4>Attivare</h4><p>Un'azione verificabile entro 15 giorni, con responsabile e scadenza.</p></div>
  <div class="pipeline-step"><div class="step">05</div><h4>Misurare</h4><p>Esito a 3, 6 e 12 mesi e confronto nel rollout.</p></div>
</div>

### I decision gate dei primi 90 giorni

Il progetto non presume che formazione, trasporto o tirocini siano sempre la risposta. Prima
di attivare i moduli più costosi, il Comune verifica la domanda:

| Evidenza raccolta | Decisione |
|---|---|
| almeno 30 esperienze retribuite con attività e mentor verificati | attivare il modulo esperienza |
| gap di competenza ricorrente associato a posizioni reali | progettare una formazione breve e mirata |
| trasporto barriera primaria e opportunità coerente già identificata | testare un supporto mobilità sul sottogruppo |
| domanda insufficiente o non verificabile | concentrare le risorse su outreach, orientamento e mercato metropolitano |

<div class="callout decision"><strong>Perché non proponiamo semplicemente un corso.</strong> I dati identificano il segmento da raggiungere, ma non dimostrano quale competenza sia richiesta dai datori. Il fabbisogno formativo deve essere verificato prima, non assunto.</div>

### Cosa misura il pilota

- **Outcome primario:** quota occupata, in istruzione o in formazione qualificante a sei mesi,
  con esito ancora attivo al dodicesimo mese.
- **Processo:** primo contatto entro 30 giorni, assessment e piano entro 15, avvio dell'azione
  concordata entro 30.
- **Qualità:** durata e tipologia del contratto, continuità dell'esito, ricaduta
  nell'inattività e coerenza dichiarata tra indirizzo di studio e attività.
- **Impatto:** rollout scaglionato; tra persone con uguale priorità, ordine casuale quando
  possibile e confronto temporaneo con chi entra più tardi.

I parametri del primo anno — **200 partecipanti, quattro case manager, 90 giorni di
preparazione** — sono ipotesi di capacità del pilota, da validare con Comune, scuole e CPI.
Non vengono presentati come stime prodotte dai dati osservazionali.
